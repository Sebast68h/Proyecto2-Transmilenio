```python
ruta = "../data/raw/validaciones/20261115.csv"

columnas = ["Fecha_Transaccion", "Estacion_Parada", "Acceso_Estacion",
            "Dispositivo", "Linea", "Nombre_Perfil", "Numero_Tarjeta"]

df = pd.read_csv(ruta, usecols=columnas)
df.shape
```

```python
catalogo = df.groupby(["Linea", "Estacion_Parada"]).size().reset_index(name="validaciones_dia")

catalogo.to_csv("../data/processed/catalogo_estaciones_20261115.csv", index=False)
catalogo[catalogo["Linea"] == "(36) Zona A Caracas"]
```
