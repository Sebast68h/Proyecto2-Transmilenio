Calle 72: abre Calle 72 y cierra Calle 76.

```python
buscar72 = calle72[calle72["estacion"].str.contains("72|76|Flores|Heroes|Polo|NQS Calle 75", case=False)]
buscar72[["codigo","estacion","linea"]].drop_duplicates()
```

Calle 53: abre Calle 53 y cierra la temporal Marly.

```python
buscar53 = calle53[calle53["estacion"].str.contains("Marly|57|Flores|39|Campin|Movistar|Nacional", case=False)]
buscar53[["codigo","estacion","linea"]].drop_duplicates()
```

Calle 26: abre Calle 26 y cierra la temporal Calle 34.
