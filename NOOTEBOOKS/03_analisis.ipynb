Con estas búsquedas anotamos los códigos de cada estación y su papel en el evento: nueva, cerrada o vecina.

```python
calle72roles= calle72.merge(roles72, on=["evento","codigo"], how="left", validate="m:1")
calle72roles.groupby("rol")["codigo"].unique()
```

```python
calle53roles= calle53.merge(roles53, on=["evento","codigo"], how="left", validate="m:1")
calle53roles.groupby("rol")["codigo"].unique()
```

```python
calle26roles= calle26.merge(roles26, on=["evento","codigo"], how="left", validate="m:1")
calle26roles.groupby("rol")["codigo"].unique()
```

Unimos cada evento con la tabla de roles para saber qué papel tiene cada estación. Las estaciones que no son del evento quedan sin rol.

```python
calle72roles.loc[(calle72roles["linea"] == "(33) Zona B AutoNorte") & (calle72roles["rol"].isna()), "rol"] = "control"
calle53roles.loc[(calle53roles["linea"] == "(33) Zona B AutoNorte") & (calle53roles["rol"].isna()), "rol"] = "control"
calle26roles.loc[(calle26roles["linea"] == "(33) Zona B AutoNorte") & (calle26roles["rol"].isna()), "rol"] = "control"
```

Control se refiere a la troncal que vamos a comparar para ver si nuestros cambios registrados ocurrieron en todo el sistema o en nuestra zona de analisis unicamente.

```python
calle26roles["rol"].value_counts(dropna=False)
```

```python
diario26 = calle26roles.groupby(["fecha","periodo","codigo","estacion","rol","hora"])["validaciones"].sum().reset_index()
diario72 = calle72roles.groupby(["fecha","periodo","codigo","estacion","rol","hora"])["validaciones"].sum().reset_index()
diario53 = calle53roles.groupby(["fecha","periodo","codigo","estacion","rol","hora"])["validaciones"].sum().reset_index()
```

```python
diario53
```

```python
estaciones72 = diario72[diario72["rol"]!= "control"]
antes72 = estaciones72[estaciones72["periodo"]== "antes"].groupby(["rol","codigo","estacion"])["validaciones"].mean().reset_index(name="antes")
despues72 = estaciones72[estaciones72["periodo"] == "después"].groupby(["rol","codigo","estacion"])["validaciones"].mean().reset_index(name="despues")

cambio72 = antes72.merge(despues72, on=["rol","codigo","estacion"], how="outer").fillna(0).round()
cambio72["cambio"] = cambio72["despues"] - cambio72["antes"]
cambio72
```

```python
estaciones53 = diario53[diario53["rol"] != "control"]
antes53 = estaciones53[estaciones53["periodo"] == "antes"].groupby(["rol","codigo","estacion"])["validaciones"].mean().reset_index(name="antes")
despues53 = estaciones53[estaciones53["periodo"] == "después"].groupby(["rol","codigo","estacion"])["validaciones"].mean().reset_index(name="despues")

cambio53 = antes53.merge(despues53, on=["rol","codigo","estacion"], how="outer").fillna(0).round()
cambio53["cambio"] = cambio53["despues"] - cambio53["antes"]
cambio53
```

```python
estaciones26 = diario26[diario26["rol"] != "control"]
antes26 = estaciones26[estaciones26["periodo"] == "antes"].groupby(["rol","codigo","estacion"])["validaciones"].mean().reset_index(name="antes")
despues26 = estaciones26[estaciones26["periodo"] == "después"].groupby(["rol","codigo","estacion"])["validaciones"].mean().reset_index(name="despues")

cambio26 = antes26.merge(despues26, on=["rol","codigo","estacion"], how="outer").fillna(0).round()
cambio26["cambio"] = cambio26["despues"] - cambio26["antes"]
cambio26
```

Hallazgos preliminares

Calle 72: los que entraban al norte de la 76 se fueron a Héroes o Polo, y los del sur de la 76 se fueron a la 72, al igual que una gran parte de los de Flores. La 72 recuperó casi toda la demanda de la 76.

Calle 53: la definitiva está más al norte que Marly, así que los que entraban al norte de Marly se fueron a la 53, y los del sur cambiaron su estación por la temporal de la 39 que es mas al sur. La 57 bajó porque ahora le queda cerca la 53.

Calle 26: todas las estaciones bajaron. Pensamos que fue por las vacaciones de mitad de año, porque la apertura fue en esa época. Lo comprobamos con la comparación con la troncal.

```python
diario72["grupo"]="zona"
diario72.loc[diario72["rol"]=="control","grupo"]="control"
diario72.groupby("grupo")["estacion"].unique()
```

Separamos las estaciones en dos grupos, por zona (las del evento) y por control (la AutoNorte)

```python
total72=diario72.groupby(["grupo","periodo","fecha"])["validaciones"].sum().reset_index()
total72["grupo"].value_counts()
```

```python
antestotal72=total72[total72["periodo"]=="antes"].groupby("grupo")["validaciones"].mean().reset_index(name="antes")
antestotal72
```

```python
despuestotal72=total72[total72["periodo"]=="después"].groupby("grupo")["validaciones"].mean().reset_index(name="despues")
despuestotal72
```

```python
comp72=antestotal72.merge(despuestotal72,on="grupo").round()
comp72["cambio_pct"]=((comp72["despues"]/comp72["antes"]-1)*100)
comp72
```

Juntamos antes y después en una tabla y calculamos cuánto cambió cada grupo en porcentaje.

```python
diario53["grupo"]="zona"
diario53.loc[diario53["rol"]=="control","grupo"]="control"
total53=diario53.groupby(["grupo","periodo","fecha"])["validaciones"].sum().reset_index()
antestotal53=total53[total53["periodo"]=="antes"].groupby("grupo")["validaciones"].mean().reset_index(name="antes")
despuestotal53=total53[total53["periodo"]=="después"].groupby("grupo")["validaciones"].mean().reset_index(name="despues")
comp53=antestotal53.merge(despuestotal53,on="grupo").round()
comp53["cambio_pct"]=((comp53["despues"]/comp53["antes"]-1)*100)
comp53
```

```python
diario26["grupo"]="zona"
diario26.loc[diario26["rol"]=="control","grupo"]="control"
total26=diario26.groupby(["grupo","periodo","fecha"])["validaciones"].sum().reset_index()
antestotal26=total26[total26["periodo"]=="antes"].groupby("grupo")["validaciones"].mean().reset_index(name="antes")
despuestotal26=total26[total26["periodo"]=="después"].groupby("grupo")["validaciones"].mean().reset_index(name="despues")
comp26=antestotal26.merge(despuestotal26,on="grupo").round()
comp26["cambio_pct"]=((comp26["despues"]/comp26["antes"]-1)*100)
comp26
```

Hallazgos de la comparación con el control

En la 72 la zona creció 2,5 % y el control 3,5 %, casi lo mismo. La 72 no trajo pasajeros nuevos, solo reacomodó a los que ya había entre la 72, Héroes y Polo.

En la 53 la zona creció 7,7 % y el control 3 %, casi 5 puntos más. La 53 sí atrajo pasajeros nuevos y no solo recibió a los de Marly.

En la 26 la zona bajó 13,4 % y el control 4,5 %. Las vacaciones explican una parte, pero la zona bajó casi 9 puntos más. Pensamos que esta zona es una zona donde se encuentran muchas universidades a su alrededor y las vacaciones la afectan más que a la AutoNorte, entonces para este evento la AutoNorte no es tan buen control.

Antes entre la 34 y la 26 había 18.372 entradas al día, y cuando cerraron la 34, la 26 sola quedó con 13.213. Pero eso no quiere decir que se perdieran usuarios, porque la 26 es una estacion principalmente que se usa para transbordo y el que cambia de troncal adentro no vuelve a validar. Puede que mucha gente de la 34 ahora entre por otra estación y llegue a la 26 por dentro. Con las validaciones no sabemos cuántas personas usan de verdad la estación.
