Con estas búsquedas anotamos los códigos de cada estación y su papel en el evento: nueva, cerrada o vecina.

```python
calle72roles= calle72.merge(roles72, on=["evento","codigo"], how="left", validate="m:1")
calle72roles.groupby("rol")["codigo"].unique()
```

```python
calle53roles= calle53.merge(roles53, on=["evento","codigo"], how="left", validate="m:1")
calle53roles.groupby("rol")["codigo"].unique()
```

Unimos cada evento con la tabla de roles para saber qué papel tiene cada estación. Las estaciones que no son del evento quedan sin rol.

```python
calle72roles.loc[(calle72roles["linea"] == "(33) Zona B AutoNorte") & (calle72roles["rol"].isna()), "rol"] = "control"
calle53roles.loc[(calle53roles["linea"] == "(33) Zona B AutoNorte") & (calle53roles["rol"].isna()), "rol"] = "control"
calle26roles.loc[(calle26roles["linea"] == "(33) Zona B AutoNorte") & (calle26roles["rol"].isna()), "rol"] = "control"
```

```python
calle26roles["rol"].value_counts(dropna=False)
```

Control se refiere a la troncal que vamos a comparar para ver si nuestros cambios registrados ocurrieron en todo el sistema o en nuestra zona de analisis unicamente.

```python
diario26 = calle26roles.groupby(["fecha","periodo","codigo","estacion","rol"])["validaciones"].sum().reset_index()
diario72 = calle72roles.groupby(["fecha","periodo","codigo","estacion","rol"])["validaciones"].sum().reset_index()
diario53 = calle53roles.groupby(["fecha","periodo","codigo","estacion","rol"])["validaciones"].sum().reset_index()
```

```python
diario53
```

```python
estaciones72 = diario72[diario72["rol"] != "control"]
antes72 = estaciones72[estaciones72["periodo"] == "antes"].groupby(["rol","codigo","estacion"])["validaciones"].mean().reset_index(name="antes")
despues72 = estaciones72[estaciones72["periodo"] == "después"].groupby(["rol","codigo","estacion"])["validaciones"].mean().reset_index(name="despues")

cambio72 = antes72.merge(despues72, on=["rol","codigo","estacion"], how="outer").fillna(0).round()
cambio72["cambio"] = cambio72["despues"] - cambio72["antes"]
cambio72
```

```python
estaciones53 = diario53[diario53["rol"] != "control"]
antes53 = estaciones53[estaciones53["periodo"] == "antes"].groupby(["rol","codigo","estacion"])["validaciones"].mean().reset_index(name="antes")
despues53 = estaciones53[estaciones53["periodo"] == "después"].groupby(["rol","codigo","estacion"])["validaciones"].mean().reset_index(name="despues")

cambio53 = antes53.merge(despues53, on=["rol","codigo","estacion"], how="outer").fillna(0).round()
cambio53["cambio"] = cambio53["despues"] - cambio53["antes"]
cambio53
```

Hallazgos preliminares

Calle 72: los que entraban al norte de la 76 se fueron a Héroes o Polo, y los del sur de la 76 se fueron a la 72, al igual que una gran parte de los de Flores. La 72 recuperó casi toda la demanda de la 76.

Calle 53: la definitiva está más al norte que Marly, así que los que entraban al norte de Marly se fueron a la 53, y los del sur cambiaron su estación por la temporal de la 39. La 57 bajó porque ahora le queda cerca la 53.

Calle 26: todas las estaciones bajaron. Pensamos que fue por las vacaciones de mitad de año, porque la apertura fue en esa época. Lo comprobamos con la comparación con la troncal.

Troncal NQS: las estaciones de la NQS casi no cambiaron. NQS Calle 75 subió apenas 1 % y Movistar Arena, El Campín y U. Nacional subieron entre 5 y 6 %. Esto muestra que con las reaperturas la gente no se cambió de troncal, sino de estación dentro de la misma zona. El pequeño aumento puede ser algo general de la ciudad, lo revisamos con la troncal de control.

```python
diario72["grupo"] = "zona"
diario72.loc[diario72["rol"] == "control", "grupo"] = "control"
diario72["grupo"].value_counts()
```

Juntamos las estaciones del evento (nueva, cerrada y vecinas) en un solo grupo llamado zona, y las de la AutoNorte en el grupo control.

```python
total72=diario72.groupby(["grupo","periodo","fecha"])["validaciones"].sum().reset_index()
total72.head()
```

Agrupamos por validaciones en cada dia

```python
antes_total72= total72[total72["periodo"] == "antes"].groupby("grupo")["validaciones"].mean().reset_index(name="antes")
antes_total72
```

```python
despues_total72= total72[total72["periodo"] == "después"].groupby("grupo")["validaciones"].mean().reset_index(name="despues")
despues_total72
```

```python
com72=antes_total72.merge(despues_total72, on="grupo").round()
com72["cambio%"] = (com72["despues"]/com72["antes"]-1)*100
com72
```
