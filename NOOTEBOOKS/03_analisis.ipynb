Hallazgos de la comparación con el control

En la 72 la zona creció 2,6 % y el control 3,6 %, casi lo mismo. La 72 no trajo pasajeros nuevos, solo reacomodó a los que ya había entre la 72, Héroes y Polo.

En la 53 la zona creció 7,7 % y el control 3 %, casi 5 puntos más. La 53 sí atrajo pasajeros nuevos y no solo recibió a los de Marly.

En la 26 la zona bajó 13,4 % y el control 4,5 %. Las vacaciones explican una parte, pero la zona bajó casi 9 puntos más. Pensamos que esta zona es una zona donde se encuentran muchas universidades a su alrededor y las vacaciones la afectan más que a la AutoNorte, entonces para este evento la AutoNorte no es tan buen control.

Antes entre la 34 y la 26 había 18.372 entradas al día, y cuando cerraron la 34, la 26 sola quedó con 13.213. Pero eso no quiere decir que se perdieran usuarios, porque la 26 es una estación que se usa principalmente para transbordo y el que cambia de troncal adentro no vuelve a validar. Puede que mucha gente de la 34 ahora entre por otra estación y llegue a la 26 por dentro. Con las validaciones no sabemos cuántas personas usan de verdad la estación.

Para comparar las tres reaperturas en un mismo gráfico pasamos las validaciones a un índice (promedio de antes = 100) y contamos los días desde la reapertura. La línea es el promedio de 5 días para ver la tendencia sin los picos de cada día.

```python
diaria=series[series["grupo"]=="zona"].copy()
diaria["fecha"]=pd.to_datetime(diaria["fecha"])
promedio_antes=diaria[diaria["periodo"]=="antes"].groupby("evento")["validaciones"].mean().reset_index(name="promedio_antes")
reapertura=diaria[diaria["periodo"]=="después"].groupby("evento")["fecha"].min().reset_index(name="reapertura")
diaria=diaria.merge(promedio_antes,on="evento",validate="m:1").merge(reapertura,on="evento",validate="m:1")
diaria["indice"]=(diaria["validaciones"]/diaria["promedio_antes"]*100).round(1)
diaria["dia"]=(diaria["fecha"]-diaria["reapertura"]).dt.days
linea_indice=px.scatter(diaria,x="dia",y="indice",color="evento",opacity=0.4,trendline="rolling",trendline_options=dict(window=5,min_periods=1),title="Tendencia de la zona: promedio de 5 días (promedio de antes = 100)")
linea_indice
```

Juntamos la Calle 76 y la Calle 72 en una sola fila para ver si la zona ganó o perdió gente. Cada línea va del promedio de antes al de después.

```python
estaciones72=cambios[cambios["evento"]=="Calle 72"].copy()
estaciones72.loc[estaciones72["rol"].isin(["cerrada","nueva"]),"estacion"]="Calle 76 → Calle 72"
estaciones72=estaciones72.groupby("estacion")[["antes","despues"]].sum().reset_index()
puntos72=pd.concat([estaciones72.assign(periodo="antes",validaciones=estaciones72["antes"]),estaciones72.assign(periodo="después",validaciones=estaciones72["despues"])])
pesas=px.scatter(puntos72,x="validaciones",y="estacion",color="periodo",title="Validaciones diarias por estación antes → después: Calle 72")
for est,antes,despues in zip(estaciones72["estacion"],estaciones72["antes"],estaciones72["despues"]):
    pesas.add_shape(type="line",x0=antes,x1=despues,y0=est,y1=est,line_color="gray",line_width=3,layer="below")
pesas.update_traces(marker_size=14)
```
