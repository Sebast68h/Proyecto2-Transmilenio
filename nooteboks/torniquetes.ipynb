```python
import pandas as pd
import plotly.express as px
import streamlit as st
```

## Datos de cada estación.

Diccionario con los datos de cada estación (calle 72, calle 53 y calle 26):

```python
estaciones = {
    "Calle 72": {
        "carpeta": "../data/processed/calle72-diarias/",
        "torniquetes_norte": 5,   
        "torniquetes_sur": 5,  
        "dias": 15,              
    },
    "Calle 26": {
        "carpeta": "../data/processed/calle26-diarias/",
        "torniquetes_norte": 5,   
        "torniquetes_sur": 5,     
        "dias": 14,              
    },
    "Calle 53": {
        "carpeta": "../data/processed/calle53-diarias/",
        "torniquetes_norte": 6,  
        "torniquetes_sur": 6,     
        "dias": 14,               
    },
}
```

## Mapa de calor.

Titulo.

```python
st.set_page_config(layout="wide")
st.title("Uso de torniquetes por acceso")
```

Definir capacidad de usuarios/validaciones por torniquete por minuto.

```python
capacidad_por_torniquete = 20 

horas = list(range(4, 24)) # tupla
```

¿Por qué esta cifra?

- La capacidad por torniquete se tomó de la guía de planeación de BRT del Instituto de Políticas para el Transporte y el Desarrollo (ITDP), según la cual un torniquete normalmente atiende entre 900 y 1.200 pasajeros por hora, es decir, entre 3 y 4 segundos por pasajero. Se usó el valor alto del rango, 1.200 pasajeros por hora (20 por minuto) para no exagerar la saturación. 

- Fuente: ITDP, The BRT Planning Guide, sección 25.7 "Station Components".


Configuración de pagina.

```python
st.set_page_config(layout="wide")
st.title("Uso de torniquetes por acceso")
```

Filtro por estacion (calle 72, calle 53, calle 26 o TODAS).

```python
eleccion = st.selectbox("Estación", ["Todas", "Calle 72", "Calle 26", "Calle 53"])

if eleccion == "Todas":
    estaciones_a_mostrar = ["Calle 72", "Calle 26", "Calle 53"]
else:
    estaciones_a_mostrar = [eleccion]
```

Bucle que lee los archivos "promedio_por_hora_sur" y "promedio_por_hora_norte" de la estacion o estaciones elegidas.
- La variable "uso" corresponde al porcentaje de la capacidad de los torniquetes de un acceso se está usando en esa hora.


```python
filas_uso = {}           # uso promedio por hora, una fila por estación y acceso
for nombre_estacion in estaciones_a_mostrar:
    estacion = estaciones[nombre_estacion]

    for acceso in ["norte", "sur"]:
        promedio = pd.read_csv(estacion["carpeta"] + "promedio_por_hora_" + acceso + ".csv")
        promedio = promedio.set_index("hora").reindex(horas) # alinear las horas de 4 a 23, aunque falten algunas en el csv

        torniquetes = estacion["torniquetes_" + acceso] # cantidad de torniquetes en el acceso
        capacidad = torniquetes * capacidad_por_torniquete #capacidad total del acceso (pasajeros por minuto)
        uso = (promedio["validaciones_promedio"] / 60) / capacidad #porcentaje de uso del acceso (promedio de pasajeros por minuto / capacidad total)
        filas_uso[nombre_estacion + " _ " + acceso.capitalize()] = uso
```

Retorna un diccionario con las llaves: *estacion/es*_sur y *estacion/es*_norte
- Cada valor es una serie de pandas con indice "hora" (4 a 23)
- Cada valor de la serie es el uso de los torniquetes del acceso en esa hora:
  la proporción de su capacidad máxima que se usa en promedio
  (0 = sin uso, 1 = al límite; puede ser mayor que 1 si la demanda supera la capacidad).

Finalmente, se grafican los datos en un mapa de calor.

```python
st.subheader("Día promedio")
matriz_uso = pd.DataFrame(filas_uso).T
grafica = px.imshow(matriz_uso, color_continuous_scale="Blues", zmin=0, zmax=1,
                    aspect="auto", text_auto=".0%",
                    labels={"x": "Hora", "y": "", "color": "Uso"})

grafica.update_layout(height=150 + 45 * len(matriz_uso),
                      margin=dict(l=10, r=10, t=10, b=10),
                      coloraxis_colorbar=dict(orientation="h", y=-0.35, tickformat=".0%"))

st.plotly_chart(grafica)

dias_texto = ", ".join(nombre + ": " + str(estaciones[nombre]["dias"]) + " días"
                       for nombre in estaciones_a_mostrar)
st.caption("Promedio de días hábiles después de la apertura de cada estación (" + dias_texto + ").")
st.caption("Uso = (validaciones promedio de la hora ÷ 60) ÷ (torniquetes fijos × 20 pasajeros por minuto). "
           "Blanco = 0 %, azul más oscuro = 100 % (al límite).")
```
